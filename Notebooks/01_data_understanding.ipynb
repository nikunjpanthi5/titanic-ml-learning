```python
bins = [0, 15, 29, 49, 64, float('inf')]

labels = [
    'Children',
    'Young Adults',
    'Adults',
    'Older Adults',
    'Seniors'
]

train['AgeGroup'] = pd.cut(
    train['Age'],
    bins=bins,
    labels=labels
)
train['AgeGroup'].value_counts()
```

```python
female_3rd = train.loc[
    (train['Sex'] == 'female') &
    (train['Pclass'] == 3)
]

female_3rd.groupby(
    ['SibSp', 'Parch']
)['Survived'].agg(['count', 'sum', 'mean'])
```

```python
family_analysis = female_3rd.groupby(
    ['SibSp', 'Parch']
)['Survived'].agg(
    count='count',
    survivors='sum',
    survival_rate='mean'
).reset_index()

family_analysis['survival_rate'] *= 100

family_analysis.sort_values(
    'count',
    ascending=False
)
```
